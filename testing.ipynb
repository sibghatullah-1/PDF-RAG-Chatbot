```python
import langchain as lc
from langchain_text_splitters import MarkdownHeaderTextSplitter
```

```python

md_docuemnt = ('docs for the rag\adaptive-layout.md')

header_to_split_on = [
    ('#','Header 1'),
    ('##','Header 2'),
    ('###','Header 3'),
]

markdown_splitter = MarkdownHeaderTextSplitter (headers_to_split_on=header_to_split_on)

with open (r"docs for the rag\adaptive-layout.md",'r',encoding='utf-8') as file:
    markdown_text = file.read()

md_header_splits = markdown_splitter.split_text(markdown_text)

md_header_splits
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter


text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 900,
    chunk_overlap = 100 ,
    length_function = len,
    separators = ['\n\n','\n',' ',''],

)

final_chunks = text_splitter.split_documents(md_header_splits)
```
